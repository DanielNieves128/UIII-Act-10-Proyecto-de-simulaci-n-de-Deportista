```python
# ==========================================
# SISTEMA DEPORTIVO IA
# Lista de diccionarios en Python
# ==========================================
# Lista vacía
deportistas = []
# Diccionario vacío
deportista = {}
while True:
    print("\n===================================")
    print("SISTEMA DEPORTIVO IA")
    print("===================================")
    print("1. Agregar deportista")
    print("2. Ver deportistas")
    print("3. Buscar deportista")
    print("4. Eliminar deportista")
    print("5. IA de entrenamiento")
    print("6. Salir")
    print("===================================")
    opcion = input("Selecciona una opción: ")
    # =====================================
    # AGREGAR DEPORTISTA
    # =====================================
    if opcion == "1":
        print("\nAGREGAR DEPORTISTA")
        nombre = input("Nombre: ")
        deporte = input("Deporte: ")
        edad = input("Edad: ")
        pais = input("País: ")
        # Diccionario vacío
        deportista = {}
        deportista["nombre"] = nombre
        deportista["deporte"] = deporte
        deportista["edad"] = edad
        deportista["pais"] = pais
        # Guardar en lista
        deportistas.append(deportista)
        print("Deportista agregado correctamente")
    # =====================================
    # VER DEPORTISTAS
    # =====================================
    elif opcion == "2":
        print("\nLISTA DE DEPORTISTAS")
        if len(deportistas) == 0:
            print(" La lista está vacía")
        else:
            numero = 1
            for d in deportistas:
                print("\n-------------------------")
                print("Deportista #", numero)
                print("Nombre :", d["nombre"])
                print("Deporte:", d["deporte"])
                print("Edad   :", d["edad"])
                print("País   :", d["pais"])
                numero = numero + 1
    # =====================================
    # BUSCAR DEPORTISTA
    # =====================================
    elif opcion == "3":
        print("\nBUSCAR DEPORTISTA")
        buscar = input("Escribe el nombre: ")
        encontrado = False
        for d in deportistas:
            if d["nombre"].lower() == buscar.lower():
                print("\nDEPORTISTA ENCONTRADO")
                print("Nombre :", d["nombre"])
                print("Deporte:", d["deporte"])
                print("Edad   :", d["edad"])
                print("País   :", d["pais"])
                encontrado = True
        if encontrado == False:
            print("No se encontró el deportista")
    # =====================================
    # ELIMINAR DEPORTISTA
    # =====================================
    elif opcion == "4":
        print("\nELIMINAR DEPORTISTA")
        eliminar = input("Nombre a eliminar: ")
        nueva_lista = []
        eliminado = False
        for d in deportistas:
            if d["nombre"].lower() != eliminar.lower():
                nueva_lista.append(d)
            else:
                eliminado = True
        deportistas = nueva_lista
        if eliminado == True:
            print("Deportista eliminado")
        else:
            print("Deportista no encontrado")
    # =====================================
    # IA DE ENTRENAMIENTO
    # =====================================
    elif opcion == "5":
        print("\nIA DE RECOMENDACIÓN")
        if len(deportistas) == 0:
            print("No hay deportistas registrados")
        else:
            for d in deportistas:
                edad = int(d["edad"])
                print("\n-------------------------")
                print("🏅", d["nombre"])
                print("Deporte:", d["deporte"])
                if edad < 18:
                    print("IA recomienda:")
                    print("Entrenamiento técnico")
                    print("Mejorar velocidad")
                    print("Practicar coordinación")
                elif edad >= 18 and edad <= 25:
                    print("IA recomienda:")
                    print("Trabajar resistencia")
                    print("Aumentar fuerza")
                    print("Participar en competencias")
                else:
                    print("IA recomienda:")
                    print("Mantener condición física")
                    print("Realizar recuperación muscular")
                    print("Ejercicios de bajo impacto")
    # =====================================
    # SALIR
    # =====================================
    elif opcion == "6":
        print("\nGracias por usar el sistema")
        print("Sigue entrenando")
        print("Programa creado por Daniel Nieves NL=45")
        break
    # =====================================
    # OPCIÓN INCORRECTA
    # =====================================
    else:
        print("Opción inválida")
```
